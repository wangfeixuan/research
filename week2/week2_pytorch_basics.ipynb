# Week 2: PyTorch Tensor and Autograd Practice

This notebook contains small exercises for tensors, matrix operations, devices, and automatic differentiation.

## Setup

Import PyTorch before starting the exercises.

```python
import torch

print("PyTorch version:", torch.__version__)
print("Done.")
```

## 1. Creating tensors

A tensor is PyTorch's basic data structure for storing numbers.

```python
# Create tensors from a list and with built-in functions
tensor_from_list = torch.tensor([1, 2, 3])
zeros = torch.zeros(2, 3)
ones = torch.ones(2, 3)

print("Tensor from list:", tensor_from_list)
print("Zeros:\n", zeros)
print("Ones:\n", ones)
print("Done.")
```

## 2. Changing tensor shapes

Reshaping changes how values are arranged without changing the number of values.

```python
# Change a tensor from shape (2, 3) to shape (3, 2)
x = torch.arange(6)
x_matrix = x.reshape(2, 3)
x_reshaped = x_matrix.reshape(3, 2)

print("Original shape:", x_matrix.shape)
print("New shape:", x_reshaped.shape)
print(x_reshaped)
print("Done.")
```

## 3. Matrix multiplication

The @ operator performs matrix multiplication.

```python
# Multiply a (2, 3) matrix by a (3, 2) matrix
A = torch.ones(2, 3)
B = torch.ones(3, 2)
C = A @ B

print("A shape:", A.shape)
print("B shape:", B.shape)
print("Result shape:", C.shape)
print(C)
print("Done.")
```

## 4. Moving tensors between devices

PyTorch can run tensors on the CPU, or on an available GPU device such as CUDA or Apple MPS.

```python
# Choose the best available device
if torch.cuda.is_available():
    device = torch.device("cuda")
elif torch.backends.mps.is_available():
    device = torch.device("mps")
else:
    device = torch.device("cpu")

small_tensor = torch.tensor([1.0, 2.0, 3.0])
small_tensor_on_device = small_tensor.to(device)

print("Selected device:", device)
print("Tensor device:", small_tensor_on_device.device)
print("Done.")
```

## 5. Creating tensors with gradients

Setting `requires_grad=True` tells PyTorch to track operations on the tensor.

```python
# Create a tensor whose gradient will be tracked
x = torch.tensor([1.0, 2.0, 3.0], requires_grad=True)

print("x:", x)
print("Requires gradient:", x.requires_grad)
print("Done.")
```

## 6. Calculating a simple function

For this exercise, use the function y = x^2 + 3x.

```python
# Calculate y element by element
y_values = x**2 + 3 * x

print("y:", y_values)
print("Done.")
```

## 7. Calling backward()

`backward()` calculates derivatives through the computation graph.
Because x has three values, sum the outputs to create one scalar before calling backward().

```python
# Turn the output into a scalar and run backpropagation
y = y_values.sum()
y.backward()

print("Scalar y:", y)
print("Done.")
```

## 8. Inspecting .grad

The derivative of y = x^2 + 3x is 2x + 3. Compare that analytical result with PyTorch's gradient.

```python
# Inspect the gradient stored in x.grad
autograd_gradient = x.grad
analytical_gradient = 2 * x.detach() + 3

print("PyTorch gradient:", autograd_gradient)
print("Analytical gradient:", analytical_gradient)
print("Gradients match:", torch.allclose(autograd_gradient, analytical_gradient))

assert torch.allclose(autograd_gradient, analytical_gradient)
print("Done.")
```

## Reflection

The main idea I learned is that PyTorch records tensor operations, and `backward()` uses that recorded computation graph to calculate gradients. The gradient is stored in `.grad`, where it can later be used by an optimizer to update model parameters.
