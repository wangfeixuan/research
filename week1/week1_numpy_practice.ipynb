##  1A: NumPy Practice

```python
import numpy as np

rng = np.random.default_rng(42)#reproducible
X = rng.normal(size=(1000, 50))

print('X:', X.shape)
```

```python
#1. Calculate the mean of every feature.
feature_mean = X.mean(axis=0)
print(feature_mean)
print(feature_mean.shape)
```

```python
#2. Calculate the variance of every feature.
feature_variance = X.var(axis=0)
print (feature_variance)
print (feature_variance.shape)
```

```python
#3. Normalize every feature.
feature_std = X.std(axis=0)
X_norm = (X - feature_mean) / feature_std 
print( X_norm.mean(axis=0)[:5]) #print 5th，the answer infinitely approaching zero
print( X_norm.std(axis=0)[:5])
```

```python
#4. Calculate the Euclidean distance between two samples.
sample_a= X[0]
sample_b =X[1]
distance = np.linalg.norm(sample_a - sample_b)
print (distance)
```

```python
#5. Calculate cosine similarity between two samples.
## cos(a,b) = (a·b) / (||a|| * ||b||)
dot_product = np.dot(sample_a,sample_b)
norm_a =np.linalg.norm(sample_a)
norm_b =np.linalg.norm(sample_b)
cosine_similarity = dot_product / (norm_a * norm_b)
print (cosine_similarity)
```

```python
#6. Calculate matrix multiplication between two compatible matrices.
A = rng.normal(size=(1000, 50))
B = rng.normal(size=(50, 20))
C = A @ B                      
print('A @ B =', C.shape)     
```

```python
# 7. Select the first 100 samples without using a loop.
first_100 = X[:100]
print(first_100)
print(first_100.shape)        
```

```python
#8. Find the sample with the largest L2 norm.（Euclidean norm）
norms = np.linalg.norm(X, axis=1)
idx_max = np.argmax(norms) 
print('Largest norm sample index:', idx_max)
print('Largest norm value:', norms[idx_max])
```

## 1BImplement Basic Functions

```python
# 1. mean_squared_error
import numpy as np
from sklearn.metrics import mean_squared_error as sk_mean_squared_error
def mean_squared_error(y_true, y_pred):
    # MSE = 1/n * sum( (y_true - y_pred)^2 )
    return np.mean((y_true - y_pred) ** 2)

y_true = np.array([1.0, 2.0, 3.0, 4.0])
y_pred = np.array([1.1, 1.9, 3.2, 4.5])

mse_result = mean_squared_error(y_true, y_pred)
lib_mse_result = sk_mean_squared_error(y_true, y_pred)

print(f"Mean Squared Error = {mse_result:.4f}")
print(f"Mean Squared Error (sklearn library) = {lib_mse_result:.4f}")
#The larger the value, the more serious the prediction was wrong.
```

```python
# 2. accuracy 
import numpy as np
from sklearn.metrics import accuracy_score as sk_accuracy_score

def accuracy(y_true, y_pred):
    # True(1)，False(0)
    return np.mean(y_true == y_pred)
# t=3 f=2  a=3/3+2 = True(1)/5 so can use mean(y_true == y_pred)
y_true = np.array([0, 1, 2, 2, 0])
y_pred = np.array([0, 2, 2, 2, 1])

acc_result = accuracy(y_true, y_pred)
lib_acc_result = sk_accuracy_score(y_true, y_pred)

print(f"Accuracy = {acc_result:.4f}")
print(f"Accuracy (sklearn library) = {lib_acc_result:.4f}")
# Accuracy represents the fraction of samples that are predicted correctly.
```

```python
# 3. sigmoid 
import numpy as np
from scipy.special import expit as sk_sigmoid

def sigmoid(z):
    # sigmoid(z) = 1 / (1 + exp(-z))
    return 1.0 / (1.0 + np.exp(-z))

z_input = np.array([-5.0, 0.0, 5.0])
sigmoid_out = sigmoid(z_input)
lib_sigmoid_out = sk_sigmoid(z_input)

print(f"sigmoid : {np.round(sigmoid_out,4)}")
print(f"sigmoid (scipy library): {np.round(lib_sigmoid_out,4)}")
# Sigmoid maps any real number into range (0,1).
```

```python
# 4. softmax 
import numpy as np
from scipy.special import softmax as lib_softmax

def softmax_simple(z):
    exp_z = np.exp(z)
    return exp_z / np.sum(exp_z, axis=-1, keepdims=True)
    
logits = np.array([2., 1., 0.1])
softmax_out = softmax_simple(logits)
lib_softmax_out = lib_softmax(logits)

print(f"softmax:\n{np.round(softmax_out,4)}")
print(f"softmax (scipy library):\n{np.round(lib_softmax_out,4)}")
print(f"sum: {np.sum(softmax_out)}")
# Softmax converts raw logits into probabilities summing to 1.
```

```python
# 5. cross_entropy 
import numpy as np
from sklearn.metrics import log_loss

# use our handwritten softmax from cell 4
def softmax_simple(z):
    exp_z = np.exp(z)
    return exp_z / np.sum(exp_z, axis=-1, keepdims=True)

def cross_entropy(y_labels, probs):
   
    n = len(y_labels)
   
    p_correct = probs[np.arange(n), y_labels]
    
    return -np.mean(np.log(p_correct + 1e-12))

logits = np.array([[2., 1., 0.1],
                   [0.5, 2., 1.]])
probs = softmax_simple(logits)
y_label = np.array([0, 1])

ce_result = cross_entropy(y_label, probs)
# add labels=[0,1,2] to fix the class number error
lib_ce_result = log_loss(y_label, probs, labels=[0,1,2])

print(y_label)
print(np.round(probs,4))
print(f"Cross Entropy Loss (my implementation) = {ce_result:.4f}")
print(f"Cross Entropy Loss (sklearn library) = {lib_ce_result:.4f}")
# Cross entropy loss becomes smaller when predicted probability of true class is higher.
```
